```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

```python
df1 = sns.load_dataset('flights')
df1
```

```python
tipcorr = df.select_dtypes(include='number')
tipcorr
```

```python
tipcorr.corr()
```

```python
sns.heatmap(tipcorr.corr(),annot=True)
```

```python
tippiv=df.pivot_table(values='tip',index='day',columns='total_bill')
tippiv
```

```python
flightpiv = df1.pivot_table(values='passengers', index='month', columns='year')
print(flightpiv)
sns.heatmap(flightpiv)
```
