```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x =np.linspace(0,5,6)
print(x)
y = np.array([3, 5, 7, 2, 4, 6])
print(y)
```

```python
a=np.arange(1,10,2)
b=a**2
print(a)
```

```python
plt.plot(x,y)
plt.title("Plot")
plt.xlabel("x axis")
plt.ylabel("y axis")

```

```python
plt.subplot(2, 2, 4)
plt.plot(x,y)
plt.subplot(2, 2, 2)
plt.plot(y,x)


plt.subplot(2,2,1)
plt.plot(a,b)
plt.subplot(2,2,3)
plt.plot(b,a)
plt.title("plot")
```

```python
fig = plt.figure(figsize=(8,2))
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x,y)
axis2=fig.add_axes([0.4,0.1,0.5,0.5])
axis2.plot(a,b)
```

```python
plt.scatter(a,b)
plt.scatter(x,y)
```

```python
plt.boxplot(x)
plt.boxplot(a)
```

```python
plt.plot(a,b)
plt.savefig("simplefig")
```
