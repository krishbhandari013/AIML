```python
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot(x=df['sex'],hue=df['size'],palette='rainbow')
```

```python
import numpy as np
sns.barplot(x=df["sex"],y=df['total_bill'],estimator=np.std)
```

```python
sns.boxplot(x=df['tip'],y=df['smoker'],palette='rainbow')
```

```python
sns.violinplot(y=df['tip'],x=df['day'],palette='rainbow')
```

```python
sns.violinplot(x=df['tip'],y=df['smoker'],palette='rainbow')
```

```python
sns.stripplot(x=df['total_bill'],y=df['day'],palette='rainbow')
```

```python
sns.swarmplot(y=df['total_bill'],x=df['day'],palette='rainbow')
```

```python
sns.swarmplot(y=df['tip'],x=df['day'],palette='rainbow')
```

```python
sns.violinplot(y= 'tip',x = 'day',data = df)
sns.swarmplot(y= 'tip',x = 'day',data = df)
```

```python
day_tip = df.groupby('day')['tip'].sum().reset_index()

sns.barplot(data=day_tip, x='day', y='tip', palette='Blues_d')
plt.title('Total tip by day')
plt.xlabel('Day')
plt.ylabel('Total tip')
plt.show()
```
