```python
import numpy as np
import pandas as pd

# ── 1. Create a messy DataFrame ──────────────────────────────────────
df = pd.DataFrame({
    'name':   ['Alice', 'Bob ', 'Charlie', 'Bob', None, 'Derek'],
    'age':    [25, 30, np.nan, 30, 22, np.nan],
    'score':  [88, 95, 120, 95, 75, np.nan],
    'team':   ['A', 'B', 'A', 'B', 'A', 'B']
})
print("=== BEFORE ===")
print(df)
print()

# ── 2. Strip whitespace & lowercase string columns ───────────────────
df['name'] = df['name'].str.strip().str.lower()

# ── 3. Remove duplicates ─────────────────────────────────────────────
df = df.drop_duplicates().reset_index(drop=True)

# ── 4. Fill missing numeric values ───────────────────────────────────
df['age']   = df['age'].fillna(df['age'].median())   # median for age
df['score'] = df['score'].fillna(df['score'].mean())  # mean for score

# ── 5. Cap outliers (score > 100 is impossible on a 0-100 scale) ────
df['score'] = df['score'].clip(upper=100)

# ── 6. Fill remaining missing name ───────────────────────────────────
df['name'] = df['name'].fillna('unknown')

print("=== AFTER ===")
print(df)
print()
print("Null counts:")
print(df.isnull().sum())   
```
